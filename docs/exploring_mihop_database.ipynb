# dataset structure
At the top level of our S3 bucket, we have a single key prefix "mihop_db" that in turn contains:

 1. The dataset license (LICENSE.md)
 2. A README.md document
 3. A Taxonomy.tsv file containing the accessions of all genomes used to construct the database and their phylogenies.
 4. A key prefixes, "mihop_db_dustmasked" containing the genomes sequences
 5. mihop_db_dustmaksed contains gzipped FASTA files containing multiple dustmasked genomes split into 1GB chunks.
 
 Full documentation for this dataset can be found at: https://github.com/mjsull/mihop

# Interacting with the dataset.

The database can be interacted with using pythons standard libraries.

This database has been designed to work with mihop, but may be usefull for processing with other metagenomic profilers. This database is not indexed so taxonomy.tsv and mihop_db_dustmasked and its contents should be downloaded to the workign directory before proceeding.

Below is an example for downloading all Escherichia genomes and writing to a file.

```python
import sys
import gzip
import glob

accession_set = set()
with open("taxonomy.tsv") as f:
    for line in if:
        if ";g__Escherichia;" in line:
            accessions_set.add(line.split("\t"))[0]


with gzip.open("Escherichia.fa.gz", "wb") as o:
    for fasta in glob.glob("mihop_db_dustmasked/*.fa.gz"):
        get_sequence = False
        with gzip.open(fasta, 'rt') as f:
            for line in f:
                if line.startswith(">"):
                    accession = line.split("|")[0][1:]
                    if accession in accessions_set:
                        get_sequence = True
                        o.write(line)
                    else:
                        get_sequence = False
                elif get_sequence:
                    o.write(line)
```

Our dataset contains a taxonomy in headerless tab separted format (.tsv).

This includes an NCBI accession in the first column and a taxonomy in the lineage string format used by NCBI.

GTDB taxonomy is used for Archaea and Bacteria and NCBI taxonomies are used for viruses and Eukaryotes.

TSV format has been used as it is human readable and allows users to easily grep taxa to determine whether they are contained in the database.

Our dataset also contains FASTA files for all genomes. The header of the FASTA files takes the following format.

>ACCESSION|contig_name description of assembly or contig.

Where ACCESSION must be present in taxonomy.tsv

contig_names must be unique to the accession or otherwise read binning will not work correctly.

This dataset can be readily downloaded using mihop.

mihop.py download_database path/to/use/for/database/

Genomes can easily be added to this database by creating a cmdd.XX.fa.gz file in mihop_db_dustmasked. Where XX can be replaced with the next number incrementally. Ensure you use the FASTA header format as stated above and add accessions and taxonomies to taxonomy.tsv

To our knowledge this dataset represents the most complete metagenomics database for clinical metagenomics.
